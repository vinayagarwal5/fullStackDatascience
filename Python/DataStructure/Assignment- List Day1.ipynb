# Python List Day 1

This notebook is a practice file for learning Python lists. It covers:
- creating empty and non-empty lists
- using `len()` to check the number of items
- adding elements with `append()`
- understanding how nested lists are stored inside a list
- checking membership with `in`
- slicing, negative slicing, and step slicing
- common list errors such as `TypeError` and `IndexError`
- working with mixed data types inside a single list


```python
# Creating an empty list
l=[]
l
```

```python
# Creating a list with one element
l=[1]
l
```

```python
# len() returns the number of items in a list
len(l)
```

```python
# Creating a list of integers
l1 = [10, 20, 30] 
l1
```

```python
# append() adds one item at the end of the list
l1.append(40)
l1.append(50)
```

```python
# The list now contains all appended values
l1
```

```python
# The length of the list is 5
len(l1)
```

```python
# append() accepts only one value at a time
l1.append(34,35)
```

```python
# The earlier invalid call did not change the list
l1
```

```python
# Appending a nested list adds the whole list as a single item
l1.append(['a','b','c','d'])
```

This list now contains a nested sublist.
The value `['a','b','c','d']` is stored as one element inside `l1`.

```python
# Displaying the list after appending a nested list
l1
```

```python
# A list can store mixed data types such as int, float, string, complex, and bool
l2= [34,7.4,'jivan',(1+2j),True]
```

```python
# Printing all lists to inspect values
print(l)
print(l1)
print(l2)
```

```python
# Checking the length of each list
print(len(l))
print(len(l1))
print(len(l2))
```

```python
# Showing the list again
l1
```

```python
# Membership test: does this exact list exist as an element?
['a', 'b', 'c','d'] in l1
```

```python
# A single list [10] is not present as one element
[10] in l1
```

```python
# List remains unchanged
l1
```

```python
# Slicing: [:] gives a copy of the full list
l1[:]
```

```python
# [:1] returns the first element only
l1[:1]
```

```python
# [:-1] excludes the last item
l1[:-1]
```

```python
# [:-3] keeps the first three items
l1[:-3]
```

```python
# [1:-1] slices from index 1 to the second-last index
l1[1:-1]
```

```python
# Accessing an index beyond the list range raises IndexError
l1[100]
```

```python
# The list is still intact after the failed access
l1
```

```python
# [4:] returns from the 5th element onward
l1[4:]
```

```python
# [0:10:3] picks elements at step 3
l1[0:10:3]
```

```python
# [0:10:4] picks elements at step 4
l1[0:10:4]
```

```python
# End index can exceed the list length; Python stops at the end
l1[0:20:4]
```

```python
# Starting index 1 and step 4 gives a sparse subset of elements
l1[1:20:4]
```

```python
# [::2] picks every second item from the start to the end
l1[::2]
```

```python
# [::3] picks every third item
l1[::3]
```

```python
# Final state of the list before counting elements
l1
```

```python
# len() counts all items, including a nested list as one element
len(l1)
```

```python
# count() tells how many times a value appears in the list
l1.count(10)
```

```python
# Step slicing with negative steps can reverse the order
l1[::-2] ## reverse order
```
