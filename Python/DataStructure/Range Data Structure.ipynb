# 🔢 Python Range Data Structure

The `range` data type is an immutable sequence used to generate a series of numbers.

It is commonly used in:
- ✔ Loops  
- ✔ Iteration  
- ✔ Index generation  

---

# 📌 Syntax

```python
range(start, stop, step)
```

This notebook demonstrates common patterns and explains how `range` behaves when converted to lists, tuples, and sets.

| Parameter | Description | Example |
|---|---|---|
| start | Starting number (inclusive). Defaults to 0 if omitted. | `range(0,5)` |
| stop | End number (exclusive). The sequence stops *before* this value. | `range(5)` |
| step | Increment between values. Defaults to 1. Can be negative to count down. | `range(0,10,2)` |

Examples below include comments explaining each operation.

````python
# range(stop): generates numbers from 0 up to stop-1
range(5)  # represents values 0,1,2,3,4
````

````python
# convert range to a list to materialize the values
list(range(5))  # [0, 1, 2, 3, 4]
````

````python
# range(start, stop, step): here step=2 selects every second number
list(range(0, 5, 2))  # [0, 2, 4]
````

````python
# create a list from range with step of 3 and print it
data_range = list(range(0, 15, 3))  # [0, 3, 6, 9, 12]
print(data_range)
````

````python
# Accessing elements: list indexing works normally once converted to list
data_range[1]  # returns the element at index 1 (3)
````

````python
# Because we created a list, we can modify its elements
data_range[1] = 4  # change the value at index 1 from 3 to 4
print(data_range)
````

````python
# range(start, stop, step) displays the range object when not materialized
range(5, 40, 2)  # represents odd-even stepping sequence from 5 to 39
````

````python
# convert range to a tuple to get an immutable sequence of the values
tuple(range(5, 40, 2))
````

````python
# convert range to a set: ordering is not guaranteed and duplicates removed (not relevant here)
set(range(5, 40, 2))
````

````python
# creating a set containing the range object itself (not the expanded values)
# this produces a set with a single element: the range object
s = {range(5, 40, 2)}
s
````

````python
# iterate over a range directly in a for-loop (no conversion needed)
for i in range(5, 10, 2):
    print(i)  # prints 5, 7, 9
````

## Conclusion and Notes

- `range` is lazy and efficient: it does not allocate all values until you materialize it (e.g., with `list()` or `tuple()`).
- Converting to a mutable sequence (like `list`) allows in-place edits.
- Using `range` directly in loops is the most memory-efficient pattern for iteration.
