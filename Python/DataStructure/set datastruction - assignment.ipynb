```python
# Define three example sets a, b, and c for later set operations
# a contains numbers 1 through 5, b overlaps with a, c is a separate small set
a={1,2,3,4,5}
b={4,5,6,7,8}
c={8,9,10}
```

```python
# Intersection of three sets a, b and c.
# The & operator finds common elements among sets.
a&b&c
```

```python
# Difference: elements in a that are not in b
# a.difference(b) returns a new set with items from a that are not present in b
a.difference(b)
```

```python
# Difference: elements in b that are not in a
# b.difference(a) shows items unique to b
b.difference(a)
```

```python
# Symmetric difference between b and c: items in either b or c but not both
# The ^ operator computes symmetric difference
b^c
```

```python
# Symmetric difference between a and c
# This returns elements that are in a or c but not in both
a^c
```

```python
# Chained symmetric differences: a ^ b ^ c
# Evaluated left to right, it yields items that appear in an odd number of sets
a^b^c
```

```python
# Define larger sets for additional checks
a2 = {1,2,3,4,5,6,7,8,9}
```

```python
# Define b2 which overlaps with a2
b2 = {3,4,5,6,7,8}
```

```python
# Display a2, b2, and c2 together (last expression in cell is returned)
a2
b2
c2
```

```python
# Check if b2 and c2 are disjoint (no common elements)
b2.isdisjoint(c2)
```

```python
# Check if a2 is a superset of b2 (i.e., a2 contains all elements of b2)
a2.issuperset(b2)
```

```python
# Elements present in a2 but not in b2
a2.difference(b2)
```
