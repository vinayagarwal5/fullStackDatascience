```python
# Creating an empty tuple
t = ()
t
```

```python
# Checking the type of the tuple object
type(t)
```

```python
# Creating a tuple with three integer values
t1 = (10, 20, 30)
```

```python
# Displaying the tuple
t1
```

```python
# Creating a tuple with mixed data types
t2 = (11, 14, 16, 'nit', True, 1 + 2j)
```

```python
# Printing all tuples to inspect values
print(t)
print(t1)
print(t2)
```

```python
# Counting how many times 10 appears in t1
t1.count(10)
```

```python
# Tuples are immutable, so append() is not allowed
t1.append(10)
```

```python
# Accessing an out-of-range index raises an IndexError
t1[20]
```

```python
# Accessing a valid index returns that element
t1[1]
```

```python
# Creating a tuple to store account details
icici = ('john', 123456, 'cizps')
```

```python
# Tuples do not support remove() because they are immutable
t1.remove(20)
```

```python
# Repeating a tuple using multiplication
t3 = t1 * 3
t3
```

```python
# Creating an empty dictionary
s = {}
s
```

```python
# Checking the type of s
type(s)
```

```python
# Creating a set with unordered numeric values
s1 = {100, 4, 10, 3, 90, 20, 50}
```

```python
# Checking the type of s1
type(s1)
```

```python
# Creating a set with mixed data types
s2 = {3, 1 + 2j, True, 3.4}
```

```python
# Printing the types of s, s1, and s2
print(type(s))
print(type(s1))
print(type(s2))
```

```python
# Adding a new value to the set
s1.add(25)
s1
```

```python
# Copying a set into a new set
s3 = s1.copy()
s3
```

```python
# Printing all sets for comparison
print(s1)
print(s2)
print(s3)
```

```python
# Dictionaries are accessed by keys, not index numbers
s[0]
```

```python
# Sets do not support indexing or slicing
s1[:]
```

```python
# pop() removes and returns one arbitrary element from a set
s1.pop()
```

```python
# discard() removes an item if it exists, otherwise it does nothing
s1.discard(1000)
```
