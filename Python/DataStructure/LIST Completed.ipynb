# Python Lists: Complete Guide

A list is one of the most commonly used data structures in Python. It stores multiple items in a single variable and keeps them in order. Lists are mutable, which means you can change their contents after creation.

In this notebook, we will learn how to create lists, access elements, add or remove values, and work with common list methods.

```python
l = []
print(l)
print(len(l))
```

## 1. Creating an Empty List

An empty list is created using square brackets `[]`. The `len()` function tells us how many items are in the list. This is the starting point for storing data in a list.

```python
l1 = [10, 20, 30]
print(l1)
print(len(l1))
```

## 2. Creating a List with Values

A list can hold multiple values at once. Each value is separated by a comma. Lists preserve the order in which values were added.

```python
l1.append(40)
l1.append(50)
l1.append(60)
l1.append(10)
print(l1)
print(len(l1))
```

## 3. Adding Elements with `append()`

The `append()` method adds a single element to the end of the list. It accepts exactly one argument. If you try to pass two values at once, Python raises a `TypeError`.

```python
# This will raise an error because append() accepts only one argument
# l1.append(70, 80)
l1.append(['a', 'b', 'c'])
print(l1)
```

## 4. Appending a List Inside a List

A list can contain any type of value, including another list. In this example, we append `['a', 'b', 'c']` as a single item in the list. The nested list is treated as one element.

```python
l2 = [34, 5.3, 'prakash', 1 + 2j, True]
print(l2)
```

## 5. Lists Can Hold Different Data Types

Python lists are flexible. A single list can contain integers, floats, strings, complex numbers, booleans, and even nested lists. This makes lists highly useful for real-world data storage.

```python
print(type(l))
print(type(l1))
print(type(l2))
print(id(l))
print(id(l1))
print(id(l2))
```

## 6. Checking Type and Memory Address

The `type()` function tells us the data type of an object. The `id()` function returns the memory location of the object in Python. This helps us understand how Python stores different list objects.

```python
print(10 in l1)
print(['a', 'b', 'c'] in l1)
print(100 in l1)
```

## 7. Membership Test with `in`

The `in` operator checks whether a value exists in a list. If the value is present, it returns `True`; otherwise it returns `False`. This is useful when searching inside a list.

```python
l3 = l2.copy()
print(l3)
print(id(l2))
print(id(l3))
l3.clear()
print(l3)
print(len(l3))
del l3
```

## 8. Copy, Clear, and Delete

The `copy()` method creates a duplicate of a list. `clear()` removes all elements from the list, while `del` deletes the variable itself. These are important operations for managing list data safely.

```python
print(l1.count(10))
print(l2.count(5.3))
print(l1[:])
print(l1[-1])
print(l1[4:])
print(l1[3:-1])
print(l1[0:10:3])
```

## 9. Counting and Slicing

`count()` returns how many times an item appears in a list. Slicing allows us to access a range of values in a list. For example, `l1[3:-1]` selects values from index 3 up to index -1, excluding the last item.

```python
# This will raise an IndexError because index 100 is out of range
# print(l1[100])
print(l1[0])
print(l1[1])
print(l1[-1])
```

## 10. Indexing and Range Errors

List indexes start from 0. Accessing an index that does not exist raises an `IndexError`. Python list indexing is powerful, but we must remain within the valid range.

```python
l = [10, 20, 30, 40, 50, 60, 10, ['a', 'b', 'c']]
print(l[0])
print(l[3:])
print(l[::2])
print(l[::-1])
```

## 11. Advanced Slicing

Slicing can be used with a start, stop, and step value. The syntax `list[start:stop:step]` is very useful for selecting a subset of elements or reversing a list.

```python
l = [10, 20, 30, 40, 50, 60, 10, ['a', 'b', 'c']]
l.insert(2, 25)
print(l)
l.pop()
print(l)
l.remove(10)
print(l)
l.sort()
print(l)
l.reverse()
print(l)
```

## 12. Common List Methods

Python provides several built-in methods for list manipulation: `insert()`, `pop()`, `remove()`, `sort()`, and `reverse()`. These methods help us update and organize the list efficiently.

## Summary

- Lists are ordered and mutable.
- They can store different data types.
- `append`, `insert`, `pop`, and `remove` modify the list.
- `len`, `count`, and `in` help inspect list data.
- Slicing helps access subsets of a list.

This notebook is a practical beginner guide to Python lists.
