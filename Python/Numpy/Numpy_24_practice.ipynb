```python
import numpy as np
```

```python
s = np.array([1,2,3,4])
mean = np.mean(s)
print("Mean of array:\n", mean)
```

```python
#variance of array
#daviation from mean 1-2.25,2-2.25,3-2.25,4-2.25
#variance of above daviation square(1-2.25+2-2.25+3-2.25+4-2.25)/4


var= np.var(s) # 
print("Variance of array:\n", var)
```

```python
#standard daviation of array
std_var=np.sqrt(var)
print("standard variation pf array through tradional methods : ",std_var)
std_var1= np.std(s)
print("Standar variation of array using numpy function: ", std_var1)
```

```python
minimum = np.min(s)
print("minimum vaue from array: ",minimum)
```

```python
maximum = np.max(s)
print("maximum vaue from array: ",maximum)
```
