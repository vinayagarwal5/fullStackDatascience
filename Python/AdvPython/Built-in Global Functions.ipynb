## Enumarate()

```python
l1 = [1,5,6,8,3]
for i in enumerate(l1):
    #print(i)               # i in interator count
    print(f"Iterator {i[0]}")
    print(f"Value {i[1]}")
```

## All() & Any()

```python
# If 0 in the list all will give False
# If empty list any will give False
l  = []
l1 = [1,5,6,8]
l2 = [True, 'a',1,2]
l3 = [5,0,7]
```

```python
print(all(l))
print(all(l1))
print(all(l2))
print(all(l3))
```

```python
print(any(l))
print(any(l1))
print(any(l2))
print(any(l3))
```
