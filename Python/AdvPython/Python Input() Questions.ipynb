### Python Input() Questions

This notebook demonstrates how to capture user input using `input()`, convert it to numeric types, and validate values such as positive, negative, and zero. All original execution outputs are preserved.

### 1. What is the input() function in Python used for?
To take the input from the user. By default it'll take input as string.

```python
name = input("Enter your name")
print(f"Welcome {name}")
```

### 2. How can you accept an integer as input from the user using input()?

```python
age = int(input("Enter your age"))
print(f"Your age {age}")
```

### 3. How do you accept a float input from the user?

```python
temp = float(input("Enter your area temparature"))
print(f"Your area temparature {temp:.2f}")
```

### 4. How can you take multiple space-separated values as input?

```python
multi = input("Enter Multiple Space Separated Value").split()

for i,val in enumerate(multi):
    print(f"Entered input {i+1}: {val}")
```

### 5. How do you check if a number entered by the user is positive, negative, or zero?

```python
number = int(input("Enter a number"))

if number < 0:
    print(f"Given number {number} is negative")
elif number > 0:
    print(f"Given number {number} is positive")
else:
    print(f"Given number {number} is zero")
```

### 6. How do you convert user input to a list of integers?

```python
nums = list(map(int, input("Enter list of numbers").split()))
print(nums)
```
