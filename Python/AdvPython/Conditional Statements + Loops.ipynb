```python
x = 8
r = x%2
if r == 0:
    print('Even Number')

print('Odd Number')
```

```python
x = 8
r = x%2
if r == 0:
    print('Even Number')
else:
    print('Odd Number')
```

```python
x = 9
r = x%2
if r == 0:
    print('Even Number')
else:
    print('Odd Number')
```

```python
x = 9
r = x%2

if r == 0:
    print('Even Number')
elif r == 1:
    print('Odd Number')
else:
    print('Not a number')
```

```python
x = 9
r = x%2

if r == 0:
    print('Even Number')
    if x > 5:
        print("Greater than 5")
elif r == 1:
    print('Odd Number')
    if x > 5:
        print("Greater than 5")
else:
    print('Not a number')
```
