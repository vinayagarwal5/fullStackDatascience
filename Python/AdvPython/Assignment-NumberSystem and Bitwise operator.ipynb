# Bitwise operators

These operators perform operations on the binary representation of integers.
Common bitwise operators in Python are `~`, `&`, `|`, `^`, `<<`, and `>>`.

```python
# Bitwise NOT (~) flips every bit, effectively giving the two's complement form
~13
```

```python
# Bitwise AND (&) keeps only the bits that are set in both numbers
12&11
```

```python
# Show the binary value of the integer 11 using bin()
bin(11)
```

```python
# Bitwise OR (|) keeps bits that are set in either number
13 | 14
```

```python
# Display the binary representation of 14
bin(14)
```

```python
# Display the binary representation of 13
bin(13)
```

```python
# Right shift (>>) moves bits to the right, effectively dividing by 2^n
13>>2
```

```python
# Left shift (<<) moves bits to the left, effectively multiplying by 2^n
13<<2
```

# XOR example

The XOR operator `^` returns 1 where bits differ and 0 where they are the same.
For example: 12 ^ 13 = 1.

```python
# Bitwise XOR (^) gives 1 when bits differ between 12 and 13
12^13
```
