```
import os
from groq import Groq
from google.colab import userdata

GROQ_API_KEY = ""
client = Groq(api_key=GROQ_API_KEY)

# The path 'D:/GIT_HUB/FSDS_PROJECT/21st - Llama Model, Ollama/file_example_MP3_700KB.mp3'
# is a local path and will not work in Colab. Please upload your audio file
# to the Colab environment (e.g., to /content/) and update the filename variable.
# Example: filename = "/content/file_example_MP3_700KB.mp3"
filename = "file_example_MP3_700KB.mp3"

# Ensure the filename is not empty before attempting to open the file
if filename:
    with open(filename, "rb") as file:
        transcription = client.audio.transcriptions.create(
          file=(filename, file.read()),
          model="whisper-large-v3",
          temperature=0,
          response_format="verbose_json",
        )
        print(transcription.text)
else:
    print("Please provide the path to your audio file in the 'filename' variable after uploading it to Colab.")
```
