```python
import os
from openrouter import OpenRouter

# Initialize the OpenRouter client with your API key
openrouter_client = OpenRouter(api_key=os.environ.get("OPENROUTER_API_KEY"))

# Stream the response
stream = openrouter_client.chat.send(
    model="nvidia/nemotron-3-ultra-550b-a55b:free",
    messages=[
        {
            "role": "user",
            "content": "How many r's are in the word 'strawberry'?"
        }
    ],
    stream=True
)

response_content = ""
for chunk in stream:
    # Check if delta and content exist, as some chunks might not have them (e.g., initial chunks or usage chunks)
    if chunk.choices and chunk.choices[0].delta and chunk.choices[0].delta.content:
        content = chunk.choices[0].delta.content
        response_content += content
        print(content, end="") # Print streamed content as it arrives

# The OpenRouter Python client usually provides usage info in the final stream object or when the stream is fully consumed.
# For streaming, usage is typically aggregated at the end or not directly available in individual chunks in the same way as JS SDK.
# If detailed token usage per chunk is crucial, the OpenRouter Python client's documentation should be consulted for the most accurate way.
print(f"\nFull streamed response: {response_content}")
```
